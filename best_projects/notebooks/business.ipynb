First off let's split the data:

```python
# Prepare to store predictions and actual values
predictions_list = []
actual_values_list = []

# Loop through each region's training and validation set
for i, (X_train, X_valid, y_train, y_valid) in enumerate(train_test_sets):
    # Train the model on the training set
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # Make predictions for the validation set
    predictions = model.predict(X_valid)
    
    # Store the predictions and actual values
    predictions_list.append(predictions)  # Store predictions
    actual_values_list.append(y_valid)  # Store actual values

    # Print some debug information
    print(f"Region {i} - Predictions: {predictions[:5]}, Actual: {y_valid[:5]}")
```

### Step 5: Calculating Risk and Profit for each Region

Let's do some bootstrapping to find the distribution of profit:

```python
def bootstrap_profit(predictions, actual_values, n_wells=500, n_iterations=1000):
    profits = []
    
    for _ in range(n_iterations):
        # Sample 500 indices
        sample_indices = np.random.choice(len(predictions), n_wells, replace=True)
        sample_predictions = predictions[sample_indices]
        sample_actuals = actual_values.iloc[sample_indices]  # Ensure correct indexing

        # Select the best 200 wells based on predictions
        best_indices = np.argsort(sample_predictions)[-200:]  # Get indices of best 200 wells
        selected_predictions = sample_predictions[best_indices]
        selected_actuals = sample_actuals.iloc[best_indices]

        # Calculate profit for selected wells
        best_wells, total_reserves, profit = select_summarize_and_calculate_profit(
            features=pd.DataFrame(selected_actuals),  # Adjust this as needed
            predictions=selected_predictions,
            actual_values=selected_actuals
        )

        profits.append(profit)

    return profits
```

```python
profits_region_0 = bootstrap_profit(predictions_list[0], y_valid_0)
profits_region_1 = bootstrap_profit(predictions_list[1], y_valid_1)
profits_region_2 = bootstrap_profit(predictions_list[2], y_valid_2)
```

Now that that is all set up, let's now display the results:

```python
# Display summary statistics for each region
for i, profits in enumerate([profits_region_0, profits_region_1, profits_region_2]):
    total_profit = np.sum(profits)  # Sum all profits for total profit
    avg_profit = np.mean(profits)  # Average profit per iteration
    lower_bound, upper_bound = np.percentile(profits, [2.5, 97.5])  # 95% CI
    
    # Convert profits to a NumPy array for comparison
    profits_array = np.array(profits)
    risk_of_loss = np.mean(profits_array < 0) * 100  # Percentage of negative profits

    print(f"Region {i}:")
    print(f"  Total Profit: {total_profit:.2f}")
    print(f"  Average Profit: {avg_profit:.2f}")
    print(f"  95% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
    print(f"  Risk of Loss: {risk_of_loss:.2f}%\n")
```
